1st step, select where you will be storing tables 

```python
%sql
USE CATALOG hive_metastore; 
```

create a table with SQL or Python

```python
%sql
CREATE TABLE employees
  (id INT, name STRING, salary DOUBLE)
```

Insert statement in SQL can be structed as below. Note that multiple SQL statements in the table, only the last INSERT statement output will be outputed. 

```python
%sql
INSERT INTO employees
VALUES 
  (1, 'John', 100000), 
  (2, 'Mary', 120000); 

INSERT INTO employees
VALUES 
  (3, 'Jane', 100000), 
  (4, 'Bob', 120000);

INSERT INTO employees
VALUES 
  (5, 'Alice', 100000);
```

SELECT statement from table 

```python
%sql
SELECT * FROM employees;
```

DESCRIBE DETAIL shows the information of the table

```python
%sql DESCRIBE DETAIL employees
```

%fs ls will list the all the file logs of the table 

```python
%fs ls 'dbfs:/user/hive/warehouse/employees'
```

```python
%sql
UPDATE employees
SET salary = salary + 100
WHERE name = 'Mary';
```

Checking the history of file log change

```python
%sql
DESCRIBE HISTORY employees
```

We can revert the state of the notebook with SELECT * FROM tablename@vX

```python
%sql
SELECT * FROM employees@v1;
```
